```python
import pandas as ps
import matplotlib.pyplot as mp

data = ps.read_csv("lab0.csv")

mp.plot(data["Size"], data["Gnome"], label = "Gnome Sort")
mp.plot(data["Size"], data["Bubble"], label = "Bubble Sort")
mp.plot(data["Size"], data["Insertion"], label = "Insertion Sort")

mp.title("3 брата - N^2")
mp.xlabel("Размер массива")
mp.ylabel("Время")
mp.legend()
mp.grid(True)
mp.show()
```
