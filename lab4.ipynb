```python
import pandas as ps
import matplotlib.pyplot as mp

data = ps.read_csv("lab4.csv")

Size = data.iloc[:, 0]
Gnome_random = data.iloc[:, 1]
Gnome_best = data.iloc[:, 2]
Gnome_worst = data.iloc[:, 3]
Bubble_random = data.iloc[:, 4]
Bubble_best = data.iloc[:, 5]
Bubble_worst = data.iloc[:, 6]
Insertion_random = data.iloc[:, 7]
Insertion_best = data.iloc[:, 8]
Insertion_worst = data.iloc[:, 9]
Merge_random = data.iloc[:, 10]
Merge_best = data.iloc[:, 11]
Merge_worst = data.iloc[:, 12]
Quick_random = data.iloc[:, 13]
Quick_best = data.iloc[:, 14]
Quick_worst = data.iloc[:, 15]
Heap_random = data.iloc[:, 16]
Heap_best = data.iloc[:, 17]
Heap_worst = data.iloc[:, 18]

fig, axes = mp.subplots(3,2, figsize = (20, 20))

axes[0, 0].plot(Size, Gnome_random, label = "Gnome_random")
axes[0, 0].plot(Size, Gnome_best, label = "Gnome_best")
axes[0, 0].plot(Size, Gnome_worst, label = "Gnome_worst")
axes[0, 0].set_title('Gnome_Sort')
axes[0, 0].legend()
axes[0 ,0].grid(True)

axes[1, 0].plot(Size, Bubble_random, label = "Bubble_random")
axes[1, 0].plot(Size, Bubble_best, label = "Bubble_best")
axes[1, 0].plot(Size, Bubble_worst, label = "Bubble_worst")
axes[1, 0].set_title('Bubble_Sort')
axes[1, 0].legend()
axes[1 ,0].grid(True)

axes[2, 0].plot(Size, Insertion_random, label = "Insertion_random")
axes[2, 0].plot(Size, Insertion_best, label = "Insertion_best")
axes[2, 0].plot(Size, Insertion_worst, label = "Insertion_worst")
axes[2, 0].set_title('Insertion_Sort')
axes[2, 0].legend()
axes[2 ,0].grid(True)

axes[0, 1].plot(Size, Merge_random, label = "Merge_random")
axes[0, 1].plot(Size, Merge_best, label = "Merge_best")
axes[0, 1].plot(Size, Merge_worst, label = "Merge_worst")
axes[0, 1].set_title('Merge_Sort')
axes[0, 1].legend()
axes[0, 1].grid(True)

axes[1, 1].plot(Size, Quick_random, label = "Quick_random")
axes[1, 1].plot(Size, Quick_best, label = "Quick_best")
axes[1, 1].plot(Size, Quick_worst, label = "Quick_worst")
axes[1, 1].set_title('Quick_Sort')
axes[1, 1].legend()
axes[1, 1].grid(True)

axes[2, 1].plot(Size, Heap_random, label = "Heap_random")
axes[2, 1].plot(Size, Heap_best, label = "Heap_best")
axes[2, 1].plot(Size, Heap_worst, label = "Heap_worst")
axes[2, 1].set_title('Heap_Sort')
axes[2, 1].legend()
axes[2, 1].grid(True)


mp.xlabel("Размер массива")
mp.ylabel("Время")
mp.legend()
mp.show()
```
