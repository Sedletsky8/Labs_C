```python
import pandas as ps
import matplotlib.pyplot as mp

data0 = ps.read_csv("lab11(O0).csv")
data1 = ps.read_csv("lab12(O1).csv")
data2 = ps.read_csv("lab13(O2).csv")
data3 = ps.read_csv("lab14(O3).csv")

mp.plot(data0["Size"], data0["Bubble"], label = "Bubble Sort O0")
mp.plot(data1["Size"], data1["Bubble"], label = "Bubble Sort O1")
mp.plot(data2["Size"], data2["Bubble"], label = "Bubble Sort O2")
mp.plot(data3["Size"], data3["Bubble"], label = "Bubble Sort O3")

mp.title("Сравнение флагов O0, O1, O2, O3")
mp.xlabel("Размер массива")
mp.ylabel("Время")
mp.legend()
mp.grid(True)
mp.show()
```
