```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

data = pd.read_csv('integrate_result.csv', skiprows=1, sep=r'\s+', names=['size', 'float_error', 'double_error'])
data.columns = data.columns.str.strip()
size = data['size']
float_error = data['float_error']
double_error = data['double_error']

plt.figure(figsize = (15, 10))
plt.loglog(size, float_error, label='float_error', marker='o', linestyle='-', color='red')
plt.loglog(size, double_error, label='double_error', marker='s', linestyle='-', color = 'blue')
plt.xlabel('Количество прямоугольников (n)', fontsize=12)
plt.ylabel('Абсолютная ошибка', fontsize=12)
plt.title('Ошибки численного интегрирования', fontsize=14)
plt.legend(fontsize=12)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
x = np.linspace(0, 1, 400)

y1 = x**2 / 2
y2 = x
y3 = np.full_like(x, 0.5)

plt.figure(figsize=(10, 6))

plt.plot(x, y1, label=r'$y = \frac{x^2}{2}$', color='red', linewidth=2)
plt.plot(x, y2, label=r'$y = x$', color='blue', linewidth=2)
plt.plot(x, y3, label=r'$y = 1$', color='green', linewidth=2)


plt.xlabel('x', fontsize=12)
plt.ylabel('y', fontsize=12)
plt.title('Графики функций', fontsize=14)

plt.grid(True, alpha=0.3)

plt.legend()

plt.tight_layout()
plt.show()
```
