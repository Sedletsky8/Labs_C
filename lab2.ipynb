```python
import pandas as ps
import matplotlib.pyplot as mp

data = ps.read_csv("lab2.csv")

mp.plot(data["Size"], data["Merge"], label = "Merge Sort")
mp.plot(data["Size"], data["Quick"], label = "Quick Sort")
mp.plot(data["Size"], data["Heap"], label = "Heap Sort")

mp.title("Я скорость(быстрее только ветер(nlogn))")
mp.xlabel("Размер массива")
mp.ylabel("Время")
mp.legend()
mp.grid(True)
mp.show()
```
