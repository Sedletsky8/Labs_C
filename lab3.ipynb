```python
import pandas as ps
import matplotlib.pyplot as mp

data = ps.read_csv("lab3.csv")
data_small = ps.read_csv("lab3(small).csv")

Size = data.iloc[:, 0]
Gnome = data.iloc[:, 1]
Bubble = data.iloc[:, 2]
Insertion = data.iloc[:, 3]
Merge = data.iloc[:, 4]
Quick = data.iloc[:, 5]
Heap = data.iloc[:, 6]

Size0 = data_small.iloc[:, 0]
Gnome0 = data_small.iloc[:, 1]
Bubble0 = data_small.iloc[:, 2]
Insertion0 = data_small.iloc[:, 3]
Merge0 = data_small.iloc[:, 4]
Quick0 = data_small.iloc[:, 5]
Heap0 = data_small.iloc[:, 6]

fig, axes = mp.subplots(1,2, figsize = (30, 15))

axes[0].plot(Size, Gnome, label = "Gnome Sort")
axes[0].plot(Size, Bubble, label = "Bubble Sort")
axes[0].plot(Size, Insertion, label = "Insertion Sort")
axes[0].plot(Size, Merge, label = "Merge Sort")
axes[0].plot(Size, Quick, label = "Quick Sort")
axes[0].plot(Size, Heap, label = "Heap Sort")
axes[0].legend()
axes[0].grid(True)

axes[1].plot(Size0, Gnome0, label = "Gnome Sort")
axes[1].plot(Size0, Bubble0, label = "Bubble Sort")
axes[1].plot(Size0, Insertion0, label = "Insertion Sort")
axes[1].plot(Size0, Merge0, label = "Merge Sort")
axes[1].plot(Size0, Quick0, label = "Quick Sort")
axes[1].plot(Size0, Heap0, label = "Heap Sort")
axes[1].legend()
axes[1].grid(True)
mp.show()
```
